```python
list=["Lamiya","Aysha","Ammu","Ananya"]
count=0;
for name in list:
    count=count+ name.count('a')
print("list of names:",list)
print("total occurance of 'a':",count)
```

```python
color_list1=["yellow","white","black","red"]
color_list2=["white","black","brown","green"]
for color in color_list1:
    if color not in color_list2:
        print(color)
print("these are the colors in color_list1 one not in color_list2")
```

```python
dictionary={1:"one",3:"three",2:"two",4:"four"}
print("original dictionary",dictionary)
print("ascending order",dict(sorted(dictionary.items())))
print("descending oreder",dict(sorted(dictionary.items(),reverse =True)))
```
