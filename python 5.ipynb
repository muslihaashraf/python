```python
n=int(input("enter a number:"))
print("factors are:")
for i in range(1,n+1):
    if n%i==0:
        print(i)
```

```python
words=input("enter words:").split(",")
longest=max(words,key=len)
print("longest word:",longest)
print("length:",len(longest))
```

```python
list1=list(map(int,input("enter first list:").split()))
list2=list(map(int,input("enter second list:").split()))
if len(list1) == len(list2):
    print("both the list have same length")
else:
    print("list have different length")
if sum(list1) == sum(list2):
    print("both the list have same sum")
else:
    print("list have different sum")    
if set(list1) & set(list2):
    print("common values:",set(list1) & set(list2))
else:
    print("no common values")
```

```python
numbers=list(map(int,input("enter integers:").split()))
for i in range(len(numbers)):
    if numbers[i]>100:
        numbers[i]='over'
print("result:",numbers)
```
