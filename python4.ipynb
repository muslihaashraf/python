```python
square=lambda a:a*a
triangle=lambda l,h:0.5*l*b 
rectangle=lambda b,h:b*h
while True:
    print("\n1.Square\n2.Triangle\n3.Rectangle\n4.Exit")
    ch=int(input("Enter your choice:"))
    if(ch==1):
        a=int(input("Enter value:"))
        print("Area of square",square(a))
    elif(ch==2):
        b=int(input("Enter the breadth:"))
        h=int(input("Enter the heigth:"))
        print("Area of triangle",triangle(b,h))
    elif(ch==3):
        b=int(input("Enter the base:"))
        h=int(input("Enter the height:"))
        print("Area of rectangle",rectangle(b,h))
    elif(ch==4):
        print("Exit")
        break
    else:
        print("Invalid choice.")
```
