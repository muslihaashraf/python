```python
from Graphics import rectangle 
from Graphics import circle
from Graphics.graphics_3D import cuboid
from Graphics.graphics_3D import sphere

while(1):
    print("1.rectangle")
    print("2.circle")
    print("3.cuboid")
    print("4.sphere")
    ch=int(input("select one ="))
    if(ch==1):
        l=int(input("length ="))
        b=int(input("breadth ="))
        print("area",rectangle.rectanglearea(l,b))
        print("perimeter",rectangle.rectperimeter(l,b))
    if(ch==2):
        r=int(input("radius ="))
        print("area",circle.circlearea(r))
        print("perimeter",circle.circlepm(r))
    if(ch==3):
        l=int(input("length ="))
        b=int(input("breadth ="))
        h=int(input("height ="))
        print("area",cuboid.cuboidarea(l,b,h))
        print("perimeter",cuboid.cuboidperimeter(l,b,h))
    if(ch==4):
        r=int(input("radius ="))
        print("area",sphere.spherearea(r))
        print("perimeter",sphere.sphereperimeter(r))
        
        
```
